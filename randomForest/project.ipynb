```python
import pandas as pd
df = pd.read_csv("steam.csv")
df.head(20)
# создаем success_ratio
df["success_ratio"] = df["positive"] / (df["positive"] + df["negative"])

# создаем target (1 = успешная, 0 = нет)
df["success"] = df["success_ratio"].apply(lambda x: 1 if x > 0.75 else 0)

df[["name", "positive", "negative", "success_ratio", "success"]].head()
```
